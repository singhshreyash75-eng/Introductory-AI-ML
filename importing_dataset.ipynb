```python
import pandas as pd

# Load dataset (you can upload your own CSV in Colab)
df = pd.read_csv('/content/sample_data/california_housing_train.csv')

# Export example
df.to_csv('/content/output.csv', index=False)

# Dataset details
print("Shape (rows, columns):", df.shape)
print("\nFirst 5 rows:\n", df.head())

print("\nTotal elements (size):", df.size)

print("\nMissing values:\n", df.isnull().sum())

# Numerical stats
print("\nSum:\n", df.sum(numeric_only=True))
print("\nMean:\n", df.mean(numeric_only=True))
print("\nMin:\n", df.min(numeric_only=True))
print("\nMax:\n", df.max(numeric_only=True))
```
