```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a larger dataset (100 students)
np.random.seed(42)  # for reproducibility

marks = np.random.randint(40, 121, 100)  # values between 40 and 120

df = pd.DataFrame({'Marks': marks})
display(df)

# Min-Max Scaling
df['MinMax'] = (df['Marks'] - df['Marks'].min()) / (df['Marks'].max() - df['Marks'].min())

# Z-score (Standard Deviation)
df['Z_std'] = (df['Marks'] - df['Marks'].mean()) / df['Marks'].std()

# Z-score (MAD)
mad = (df['Marks'] - df['Marks'].mean()).abs().mean()
df['Z_MAD'] = (df['Marks'] - df['Marks'].mean()) / mad

# Decimal Scaling
j = len(str(int(df['Marks'].abs().max())))
df['Decimal'] = df['Marks'] / (10 ** j)

display(df)

plt.figure()

plt.plot(df['Marks'], df['MinMax'], label='Min-Max Scaling')
plt.plot(df['Marks'], df['Z_std'], label='Z-score (Std)')
plt.plot(df['Marks'], df['Z_MAD'], label='Z-score (MAD)')
plt.plot(df['Marks'], df['Decimal'], label='Decimal Scaling')

plt.xlabel("Marks")
plt.ylabel("Scaled Values")
plt.title("Feature Scaling Comparison")

plt.legend()
plt.grid()

plt.show()

def split_data(data, train_ratio, val_ratio, test_ratio):
    shuffled = data.sample(frac=1, random_state=42).reset_index(drop=True)
    n = len(data)

    train_end = round(train_ratio * n)
    val_end = train_end + round(val_ratio * n)

    train = shuffled[:train_end]
    val = shuffled[train_end:val_end]
    test = shuffled[val_end:]

    return train, val, test

splits = [
    (0.7, 0.2, 0.1),
    (0.7, 0.1, 0.2),
    (0.6, 0.2, 0.2)
]

for i, (tr, va, te) in enumerate(splits, 1):
    train, val, test = split_data(df, tr, va, te)

    print(f"\nSplit {i}: Train={tr*100}%, Validation={va*100}%, Test={te*100}%")

    print("\nTrain Data:")
    display(train)

    print("\nValidation Data:")
    display(val)

    print("\nTest Data:")
    display(test)
```
