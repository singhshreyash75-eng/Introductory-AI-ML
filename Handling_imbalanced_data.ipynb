```python
import pandas as pd
from sklearn.utils import resample

# Create dummy imbalanced dataset
df = pd.DataFrame({
    'feature': range(100),
    'target': [0]*90 + [1]*10
})

# Separate classes
majority = df[df.target == 0]
minority = df[df.target == 1]

# Upsample minority
minority_upsampled = resample(minority,
                             replace=True,
                             n_samples=len(majority),
                             random_state=42)

df_balanced = pd.concat([majority, minority_upsampled])

print("Before:\n", df['target'].value_counts())
print("\nAfter:\n", df_balanced['target'].value_counts())
```
